```python
import pandas as pd
import numpy as np
from IPython.display import display

print('Pandas version:', pd.__version__)
```

```python
rng = np.random.default_rng(11)
n = 30
names = ['Aarav', 'Diya', 'Rohan', 'Sneha', 'Karthik', 'Priya', 'Vikram', 'Ananya', 'Arjun', 'Meera',
         'Rahul', 'Kavya', 'Nikhil', 'Isha', 'Suresh', 'Lakshmi', 'Manoj', 'Divya', 'Harsh', 'Pooja',
         'Varun', 'Neha', 'Gautam', 'Swati', 'Ajay', 'Ritu', 'Deepak', 'Anjali', 'Sanjay', 'Tara']
departments = rng.choice(['HR', 'IT', 'Finance', 'Sales', 'Marketing'], n)
experience = rng.integers(1, 16, n)
base = {'HR': 30000, 'IT': 50000, 'Finance': 42000, 'Sales': 35000, 'Marketing': 38000}
salary = np.array([base[d] for d in departments]) + experience * 2500 + rng.integers(-3000, 3000, n)

df = pd.DataFrame({
    'employee_id': np.arange(101, 101 + n),
    'name': names,
    'department': departments,
    'city': rng.choice(['Chennai', 'Bengaluru', 'Hyderabad', 'Mumbai'], n),
    'experience_years': experience,
    'salary': salary,
    'rating': rng.choice([3.0, 3.5, 4.0, 4.5, 5.0], n)})

df.to_csv('employees.csv', index=False)
df = pd.read_csv('employees.csv')
print('Total records:', len(df))
display(df.head(10))
```

```python
# ---------- Check the dataset ----------
print('Shape:', df.shape)
print('\nData types:')
print(df.dtypes)
print('\nMissing values:', df.isnull().sum().sum())

# ---------- Part A: Ascending order ----------
print('\n=== Example 1: Numerical column, ascending (salary) ===')
asc_salary = df.sort_values('salary')
display(asc_salary.head(10))

print('\n=== Example 2: Categorical column, ascending (name A to Z) ===')
asc_name = df.sort_values('name')
display(asc_name[['employee_id', 'name', 'department', 'salary']].head(10))

# ---------- Part B: Descending order ----------
print('\n=== Example 3: Numerical column, descending (salary) ===')
desc_salary = df.sort_values('salary', ascending=False)
display(desc_salary.head(10))

print('\n=== Example 4: Categorical column, descending (name Z to A) ===')
desc_name = df.sort_values('name', ascending=False)
display(desc_name[['employee_id', 'name', 'department', 'salary']].head(10))

# ---------- Part C: Multiple columns ----------
print('\n=== Example 5: Two columns, both ascending (department, salary) ===')
multi_asc = df.sort_values(['department', 'salary'])
display(multi_asc[['name', 'department', 'salary']].head(12))

print('\n=== Example 6: Two columns, mixed order (department A-Z, salary high to low) ===')
multi_mixed = df.sort_values(['department', 'salary'], ascending=[True, False])
display(multi_mixed[['name', 'department', 'salary']].head(12))

print('\n=== Example 7: Three columns (department, city, experience) ===')
multi3 = df.sort_values(['department', 'city', 'experience_years'], ascending=[True, True, False])
display(multi3[['name', 'department', 'city', 'experience_years']].head(12))

# ---------- Part D: Useful extras ----------
print('\n=== Example 8: Reset row numbers with ignore_index=True ===')
clean = df.sort_values('salary', ascending=False, ignore_index=True)
display(clean.head())

print('\n=== Example 9: Top 5 and bottom 5 ===')
print('Top 5 salaries:')
display(df.nlargest(5, 'salary')[['name', 'department', 'salary']])
print('Lowest 5 salaries:')
display(df.nsmallest(5, 'salary')[['name', 'department', 'salary']])

print('\n=== Example 10: Sort a single column and sort by index ===')
print('Departments sorted A to Z (unique):', sorted(df['department'].unique()))
display(df['salary'].sort_values(ascending=False).head())
display(asc_salary.sort_index().head())   # back to the original order

print('\n=== Example 11: The original DataFrame is NOT changed ===')
print('Original first 3 employee_ids :', df['employee_id'].head(3).tolist())
print('Sorted by salary, first 3 ids  :', asc_salary['employee_id'].head(3).tolist())

# ---------- Observations ----------
top = df.sort_values('salary', ascending=False).iloc[0]
low = df.sort_values('salary').iloc[0]
print('\nOBSERVATIONS')
print('-' * 50)
print(f"1. Highest salary: {top['name']} ({top['department']}) earns {top['salary']:,}.")
print(f"2. Lowest salary: {low['name']} ({low['department']}) earns {low['salary']:,}.")
print('3. Sorting by two columns groups the data (department) and then orders it inside each group (salary).')
```
